```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Data from your table
x = np.array([1, 2, 3, 4, 5, 6, 7])
y = np.array([1.0, 1.3, 1.6, 1.9, 2.1, 2.2, 2.5])

# 2. Parameters
m = 0          # Slope
b = 0          # Intercept
L = 0.01       # Learning Rate
epochs = 1001  # Number of steps
n = float(len(x))

# Create a list to store our progress for the table
history = []

print(f"{'Step':<10} | {'m':<10} | {'b':<10} | {'MSE':<10}")
print("-" * 50)

# 3. Gradient Descent Loop
for i in range(epochs):
    # Current Prediction
    y_hat = m * x + b

    # Calculate Mean Squared Error (MSE)
    mse = np.mean((y - y_hat)**2)

    # Print progress every 100 steps
    if i % 100 == 0:
        print(f"{i:<10} | {m:<10.4f} | {b:<10.4f} | {mse:<10.6f}")
        history.append([i, m, b, mse])

    # Calculate Gradients
    # Derivative of MSE with respect to m and b
    dm = (2/n) * np.sum(x * (y_hat - y))
    db = (2/n) * np.sum(y_hat - y)

    # Update m and b
    m = m - L * dm
    b = b - L * db

# 4. Final results display
print("-" * 50)
print(f"Final Results after {epochs-1} steps:")
print(f"m = {m:.4f}")
print(f"b = {b:.4f}")
print(f"Final Equation: Y = {m:.4f}x + {b:.4f}")

# 5. Optional: Plotting the final line
plt.scatter(x, y, color='red', label='Actual Data')
plt.plot(x, m*x + b, color='blue', label='Final Regression Line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Final Linear Regression Result')
plt.legend()
plt.show()
```
